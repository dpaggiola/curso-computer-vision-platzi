```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
imagen = cv2.imread('cara_perro.jpg')
```

```python
plt.imshow(imagen)
```

```python
contorno = cv2.Canny(image=imagen, threshold1=100, threshold2=100)
plt.imshow(contorno)
```

```python
imagen = cv2.blur(imagen, ksize=(5,5))
```

```python
contorno = cv2.Canny(image=imagen, threshold1=100, threshold2=100)
plt.imshow(contorno)
```
